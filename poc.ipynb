```python
import random
import shutil
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.image as mpimg


def create_image_grid(
    input_dir: str,
    output_file: str = "grid_result.png",
    rows: int = 4,
    cols: int = 6,
    figsize: tuple = (18, 12),
    dpi: int = 150,
    seed: int | None = None,
) -> list[Path]:
    """
    Случайно выбирает rows*cols PNG-файлов, строит сетку и сохраняет её.
    ВОЗВРАЩАЕТ список выбранных файлов — чтобы их можно было использовать дальше.
    """
    input_path = Path(input_dir)
    all_png_files = sorted(input_path.glob("*.png"))

    expected = rows * cols
    if len(all_png_files) < expected:
        raise ValueError(
            f"Найдено только {len(all_png_files)} PNG-файлов, "
            f"а нужно минимум {expected}"
        )

    rng = random.Random(seed)
    png_files = rng.sample(all_png_files, expected)

    fig, axes = plt.subplots(rows, cols, figsize=figsize, dpi=dpi)
    axes = axes.flatten()

    for ax, img_path in zip(axes, png_files):
        img = mpimg.imread(img_path)
        ax.imshow(img)
        ax.set_title(img_path.stem, fontsize=10)
        ax.axis("off")

    plt.tight_layout()
    fig.savefig(output_file, bbox_inches="tight")
    plt.close(fig)
    print(f"Сетка сохранена в: {output_file}")

    return png_files          # <-- ключевой момент


def pick_random_image(
    png_files: list[Path],
    output_file: str = "secret_image.png",
    seed: int | None = None,
) -> None:
    """
    Случайно выбирает одну картинку из переданного списка и сохраняет её
    под нейтральным именем. Источник нигде не раскрывается.
    """
    if not png_files:
        raise ValueError("Список png_files пуст — нечего выбирать")

    rng = random.Random(seed)
    chosen = rng.choice(png_files)
    shutil.copyfile(chosen, output_file)

    print(f"Случайная картинка сохранена в: {output_file}")


if __name__ == "__main__":
    # 1) Строим сетку и получаем те самые 24 файла, что на ней изображены
    grid_files = create_image_grid(
        input_dir="characters",
        output_file="grid.png",
        rows=4,
        cols=6,
        seed=None,
    )

    # 2) Выбираем одну из них
    pick_random_image(
        png_files=grid_files,
        output_file="secret_image.png-",
        seed=None,
    )

    pick_random_image(
        png_files=grid_files,
        output_file="secret_image-2.png-",
        seed=None,
    )
```
